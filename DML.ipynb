##Q1

```python
print("1. Cleaning")
print("2. Normalization")
print("3. Integration")

ch = int(input("Enter choice: "))

if ch == 1:
    a = input("Enter data: ").split()
    a = [x for x in a if x.upper() != "NA"]
    print("Cleaned data:", a)

elif ch == 2:
    a = list(map(float, input("Enter data: ").split()))
    mn, mx = min(a), max(a)
    b = [(x-mn)/(mx-mn) for x in a]
    print("Normalized data:", b)

elif ch == 3:
    a = input("Enter first data: ").split()
    b = input("Enter second data: ").split()
    print("Integrated data:", a + b)

else:
    print("Invalid choice")
```

```python
from math import log2

a = input("Enter attribute values: ").split()
y = input("Enter target values: ").split()

def entropy(x):
    n = len(x)
    return -sum((x.count(v)/n)*log2(x.count(v)/n) for v in set(x))

e = entropy(y)
we = sum((a.count(v)/len(a))*entropy([y[i] for i in range(len(y)) if a[i] == v]) for v in set(a))

print("Entropy:", round(e,3))
print("Weighted Entropy:", round(we,3))
print("Information Gain:", round(e-we,3))
```

```python
print("1. Horizontal")
print("2. Vertical")
print("3. Round Robin")
print("4. Hash Based")

ch = int(input("Enter choice: "))

if ch == 1:
    r = int(input("Enter number of rows: "))
    a = [input("Enter row: ").split() for _ in range(r)]
    n = int(input("Enter number of partitions: "))

    for i in range(n):
        print("Partition", i+1, ":", a[i::n])

elif ch == 2:
    r = int(input("Enter number of rows: "))
    a = [input("Enter row: ").split() for _ in range(r)]
    n = int(input("Split after column: "))

    print("Partition 1:", [row[:n] for row in a])
    print("Partition 2:", [row[n:] for row in a])

elif ch == 3:
    a = input("Enter data: ").split()
    n = int(input("Enter number of partitions: "))

    for i in range(n):
        print("Partition", i+1, ":", a[i::n])

elif ch == 4:
    a = input("Enter data: ").split()
    n = int(input("Enter number of partitions: "))

    for x in a:
        print(x, "-> Partition", hash(x) % n + 1)

else:
    print("Invalid choice")
```

```python
from sklearn.tree import DecisionTreeClassifier

n = int(input("Enter number of records: "))

X = []
y = []

for i in range(n):
    a = list(map(int, input("RunnyNose Cough Temperature: ").split()))
    X.append(a)
    y.append(input("Disease (Flu/Fever): "))

model = DecisionTreeClassifier()
model.fit(X, y)

a = list(map(int, input("Enter new patient data: ").split()))
print("Prediction:", model.predict([a])[0])
```

```python
n = int(input("Enter number of transactions: "))

data = []
for i in range(n):
    data.append(input("Enter items: ").split())

minsup = int(input("Enter minimum support: "))

count = {}

for t in data:
    for item in t:
        count[item] = count.get(item, 0) + 1

print("Frequent items:")
for item in count:
    if count[item] >= minsup:
        print(item, ":", count[item])
```

```python
n = int(input("Enter number of records: "))

data = []
for i in range(n):
    data.append(input("Enter Product Month Sales: ").split())

print("\n1. Roll-up")
print("2. Drill-down")
print("3. Slice")
print("4. Dice")

ch = int(input("Enter choice: "))

if ch == 1:
    print("Total Sales:", sum(int(x[2]) for x in data))

elif ch == 2:
    p = input("Enter product: ")
    for x in data:
        if x[0] == p:
            print(x)

elif ch == 3:
    m = input("Enter month: ")
    for x in data:
        if x[1] == m:
            print(x)

elif ch == 4:
    p = input("Enter product: ")
    m = input("Enter month: ")
    for x in data:
        if x[0] == p and x[1] == m:
            print(x)

else:
    print("Invalid choice")
```

```python
n = int(input("Enter number of transactions: "))

data = []
for i in range(n):
    data.append(input("Enter items: ").split())

minsup = int(input("Enter minimum support: "))

items = set()
for t in data:
    for x in t:
        items.add(x)

print("Frequent items:")
for x in items:
    count = sum(x in t for t in data)

    if count >= minsup:
        print(x, ":", count)
```
